```python
import nltk #natural language ttolkit
from nltk.corpus import stopwords #removing common stopwords like and etc
from nltk.tokenize import word_tokenize#tokenize the data into individual words
from nltk.probability import FreqDist#analysing word frequencies
```

```python
#downloading nltk data
nltk.download("vader_lexicon")#used for analysing sentiments in social media tools
nltk.download("stopwords")
nltk.download("punkt")
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
sidee=SentimentIntensityAnalyzer()
```

```python

text=input("enter the text to detect emotion")#sample text
#function to detecct emotion in text
def detect_emotion(text):
     #analyze sentiment
    scores=sidee.polarity_scores(text)
    #display sentiment scores
    print("sentiment Scores:",scores)
    #determine emotion based on scores
    if scores["compound"]>=0.5:
        emotion="joy"
    elif scores["compound"]<= -0.5:
        emotion="Sadness"
    elif scores["neg"]>0.5:
        emotion="Anger"
    elif scores["neu"]>0.7:
        emotion="neutral"
    else:
        emotion="mixed"
    return emotion
emotion=detect_emotion(text)
print("Detected Emotion:",emotion)
```
